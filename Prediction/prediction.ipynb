```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

data=pd.read_csv("house_dataset.csv", engine='python', on_bad_lines='skip')
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
data['zip']=data["zip"].fillna(data["zip"].median())
data["street"]=data["street"].fillna(0)
data["br"]=data["br"].fillna(data["br"].median())
data["lsqft"]=data["lsqft"].fillna(data["lsqft"].median())
data["bsqft"]=data["bsqft"].fillna(data["bsqft"].median())
data["year"]=data["year"].fillna(data["year"].median())
#data["datesold"]=data["datesold"].fillna(data['date'])
#data.columns
#data.shape
#print(data.isnull().sum())
x=data[["county","city","zip","br","lsqft","bsqft"]]
y=data["price"]
x=pd.get_dummies(x,columns=["county","city"],drop_first=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train, y_train)
y_pred=model.predict(x_test)
print(y_pred)
print(y_test)
```
